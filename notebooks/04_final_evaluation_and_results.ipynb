```python
# Import all libraries needed for evaluation and saving the model
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
```

```python
# Select GPU if available, otherwise CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
```

```python
from src.models.custom_cnn import CustomCNN
```

```python
# Define dataset paths
DATA_DIR = os.path.join("data", "tiny-imagenet-200")
VAL_DIR = os.path.join(DATA_DIR, "val")

print("Validation directory exists:", os.path.exists(VAL_DIR))
```

```python
# Define transforms for validation data
val_transforms = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
])

# Load validation dataset
val_dataset = datasets.ImageFolder(VAL_DIR, transform=val_transforms)
val_loader = DataLoader(val_dataset, batch_size=64, shuffle=False)

print("Validation samples:", len(val_dataset))
```

```python
# Initialize model and load trained weights
model = CustomCNN(num_classes=200).to(device)

# Load saved weights (from Notebook 03)
model.load_state_dict(torch.load("custom_cnn_tiny_imagenet.pth", map_location=device))

model.eval()
print("Model loaded successfully")
```

```python
# Function to evaluate model performance
def evaluate(model, loader, device):
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, preds = torch.max(outputs, 1)

            correct += (preds == labels).sum().item()
            total += labels.size(0)

    return correct / total
```

```python
# Evaluate model on validation dataset
val_accuracy = evaluate(model, val_loader, device)
print(f"Final Validation Accuracy: {val_accuracy * 100:.2f}%")
```

```python
# Save final trained model
torch.save(model.state_dict(), "custom_cnn_tiny_imagenet_final.pth")
print("Final model saved successfully")
```
