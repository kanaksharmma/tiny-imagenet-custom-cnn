```python
import os
import shutil

base_dir = "data/tiny-imagenet-200/val"
images_dir = os.path.join(base_dir, "images")
annotations_file = os.path.join(base_dir, "val_annotations.txt")

# Read annotations
with open(annotations_file, "r") as f:
    annotations = f.readlines()

# Create class folders and move images
for line in annotations:
    img_name, class_id = line.split()[:2]
    class_dir = os.path.join(base_dir, class_id)
    os.makedirs(class_dir, exist_ok=True)

    src_path = os.path.join(images_dir, img_name)
    dst_path = os.path.join(class_dir, img_name)

    if os.path.exists(src_path):
        shutil.move(src_path, dst_path)

# Remove empty images directory
shutil.rmtree(images_dir)
```
