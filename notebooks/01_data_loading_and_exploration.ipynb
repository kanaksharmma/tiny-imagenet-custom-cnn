```python
import os

PROJECT_ROOT = os.getcwd()
DATA_DIR = os.path.join(PROJECT_ROOT, "data", "tiny-imagenet-200")

print("Project root:", PROJECT_ROOT)
print("Data dir exists:", os.path.exists(DATA_DIR))
print("Train dir:", os.listdir(os.path.join(DATA_DIR, "train"))[:5])
print("Val dir:", os.listdir(os.path.join(DATA_DIR, "val"))[:5])
```

```python
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
```

```python
train_transforms = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
])

val_transforms = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
])
```

```python
train_dataset = datasets.ImageFolder(
    root=os.path.join(DATA_DIR, "train"),
    transform=train_transforms
)

val_dataset = datasets.ImageFolder(
    root=os.path.join(DATA_DIR, "val"),
    transform=val_transforms
)
```

```python
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True, num_workers=2)
val_loader = DataLoader(val_dataset, batch_size=64, shuffle=False, num_workers=2)
```

```python
print("Train samples:", len(train_dataset))
print("Val samples:", len(val_dataset))
print("Classes:", len(train_dataset.classes))
```

```python
images, labels = next(iter(train_loader))

plt.figure(figsize=(8,8))
for i in range(9):
    plt.subplot(3,3,i+1)
    plt.imshow(images[i].permute(1,2,0))
    plt.axis("off")
plt.show()
```
